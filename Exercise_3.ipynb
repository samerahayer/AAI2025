# Exercise 3: Self-Reflection Prompt for Improving Output

**Goal:** Ask the AI to critique and improve its own summary.

**Tool:** Python in Google Colab (prompt design created with ChatGPT).


## Initial Summary Prompt

```text
Summarize the following paragraph in 2-3 sentences using clear, student-friendly language. Focus on the main idea and avoid unnecessary details.
```


## Self-Reflection Prompt

```text
Critique the summary you just created. Identify any missing main idea, unclear wording, repetition, or unnecessary detail. Then rewrite the summary so it is more accurate, concise, and easy to understand. Return both a short critique and an improved summary.
```


````python
source_text = (
    "Prompt engineering is the process of designing instructions that guide an AI system toward a useful response. "
    "Effective prompts often provide context, define a role or task, state constraints, and specify the desired output format. "
    "Prompt techniques such as chaining, ReACT, and self-reflection can improve the quality and reliability of AI-generated results."
)

summary_prompt = '''Summarize the following paragraph in 2-3 sentences using clear, student-friendly language. Focus on the main idea and avoid unnecessary details.'''
reflection_prompt = '''Critique the summary you just created. Identify any missing main idea, unclear wording, repetition, or unnecessary detail. Then rewrite the summary so it is more accurate, concise, and easy to understand. Return both a short critique and an improved summary.'''

# Simulated first-pass AI summary
initial_summary = (
    "Prompt engineering means giving an AI clear instructions so it can produce a useful response. "
    "Different prompting methods can improve AI results."
)

# Simulated self-reflection step
critique = (
    "The summary is clear, but it leaves out important details about what makes a prompt effective, "
    "such as context, constraints, and output format. It can be improved by briefly including those elements."
)

improved_summary = (
    "Prompt engineering is the process of giving an AI clear, structured instructions to help it produce useful responses. "
    "Strong prompts can include context, task requirements, constraints, and an output format, while techniques like prompt chaining, ReACT, and self-reflection can make results more reliable."
)

print("ORIGINAL TEXT:")
print(source_text)
print("\nINITIAL SUMMARY:")
print(initial_summary)
print("\nSELF-CRITIQUE:")
print(critique)
print("\nIMPROVED SUMMARY:")
print(improved_summary)

````

## Why self-reflection helps
The first response is treated as a draft. A second prompt asks the model to evaluate weaknesses and revise the output, which can improve completeness, clarity, and accuracy.
