# Exercise 2: Code Generation with ReACT Prompting

**Goal:** Use a clear reason/plan → code → run → observe → fix/verify cycle to generate and execute Python code.

**Tools:** ChatGPT for the ReACT prompt and Python solution design; Google Colab for execution.

## Full ReACT Prompt
```text
Use this ReACT-style cycle to solve the Python task below. Keep the planning concise and separate each stage with the exact labels PLAN, CODE, OBSERVE, FIX/VERIFY, and FINAL.

Task: Write a Python function named average_order_value(orders) that accepts a list of numeric customer order amounts and returns the arithmetic mean rounded to two decimals.

Requirements and constraints:
- Input: a Python list containing integers and/or floats.
- Output: a float rounded to two decimal places.
- If the list is empty, return 0.0.
- If any item is not an int or float, raise TypeError with a clear message.
- Use only Python built-ins; do not import third-party libraries.
- Show the function separately from the tests.
- Test one normal list, an empty list, and one invalid list.

Cycle:
1. PLAN: Briefly describe the approach and edge cases.
2. CODE: Generate the Python function and tests.
3. OBSERVE: Run the code and inspect all test results.
4. FIX/VERIFY: If any test fails, revise and rerun. If all tests pass, explain what was verified.
5. FINAL: State whether the solution satisfies every requirement.
```

## Iteration Note
An earlier version only used *Thought → Action → Observation → Final*. I refined it so the stages explicitly match an iterative coding workflow and added input/output requirements, allowed libraries, error handling, three test cases, and a fix-or-verify stage.

````python
react_prompt = """Use this ReACT-style cycle to solve the Python task below. Keep the planning concise and separate each stage with the exact labels PLAN, CODE, OBSERVE, FIX/VERIFY, and FINAL.

Task: Write a Python function named average_order_value(orders) that accepts a list of numeric customer order amounts and returns the arithmetic mean rounded to two decimals.

Requirements and constraints:
- Input: a Python list containing integers and/or floats.
- Output: a float rounded to two decimal places.
- If the list is empty, return 0.0.
- If any item is not an int or float, raise TypeError with a clear message.
- Use only Python built-ins; do not import third-party libraries.
- Show the function separately from the tests.
- Test one normal list, an empty list, and one invalid list.

Cycle:
1. PLAN: Briefly describe the approach and edge cases.
2. CODE: Generate the Python function and tests.
3. OBSERVE: Run the code and inspect all test results.
4. FIX/VERIFY: If any test fails, revise and rerun. If all tests pass, explain what was verified.
5. FINAL: State whether the solution satisfies every requirement."""

print("PLAN")
print("Validate that every item is numeric, handle an empty list before division, compute the mean, round to two decimals, and test normal, empty, and invalid inputs.")

print("\nCODE")
def average_order_value(orders):
    if not isinstance(orders, list):
        raise TypeError("orders must be provided as a list")
    if not orders:
        return 0.0
    if any(not isinstance(value, (int, float)) or isinstance(value, bool) for value in orders):
        raise TypeError("every order amount must be an int or float")
    return round(sum(orders) / len(orders), 2)

normal_orders = [24.99, 45.50, 30.00, 59.99]
print("Function defined successfully.")

print("\nOBSERVE")
print("Normal list result:", average_order_value(normal_orders))
print("Empty list result:", average_order_value([]))
try:
    average_order_value([20.0, "30.0"])
except TypeError as exc:
    print("Invalid list result:", type(exc).__name__ + ":", exc)

print("\nFIX/VERIFY")
assert average_order_value(normal_orders) == 40.12
assert average_order_value([]) == 0.0
try:
    average_order_value([20.0, "30.0"])
    raise AssertionError("Invalid data should have raised TypeError")
except TypeError:
    pass
print("All three tests passed, so no code revision was required after execution.")

print("\nFINAL")
print("The function satisfies the input/output, rounding, empty-list, error-handling, and library constraints.")
````

The output cell shows the normal calculation, empty-list behavior, invalid-input error handling, assertion-based verification, and final confirmation. If a test had failed, the FIX/VERIFY stage would be where the code was revised and rerun.
